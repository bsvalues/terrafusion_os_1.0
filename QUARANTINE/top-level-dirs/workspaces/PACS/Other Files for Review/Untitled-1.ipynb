```python
def get_embeddings(texts):
    response = openai.Embedding.create(
        input=texts,
        model="text-embedding-ada-002"
    )
    embeddings = [res['embedding'] for res in response['data']]
    
    for i, embedding in enumerate(embeddings):
        print(f"Embedding for document {i + 1}: {embedding[:10]}...")  # Print first 10 values
    return embeddings

documents = [
    "Real estate property values have been increasing in Benton County.",
    "The property tax rates are determined based on the assessed value.",
    "Parcel data shows a significant rise in land values."
]

embeddings = get_embeddings(documents)
```
